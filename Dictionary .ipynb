```python
#What is a Dictionary? 
#A built-in data type used to store data in the form of key-value pairs.
#Each key is unique.Each key has a corresponding value.
#It is mutable, which means we can add, update, and delete elements.
#It can written using curly braces {}.

#What is key value? 
#A Python dictionary stores data in the form of Key-Value pairs:-
#Key: A key is used to identify or access a particular value.
#Value: A value is the actual data stored under a key.

#Why do we use Dictionary?
#To store data in key-value pairs. To access values easily using their keys.
#To store related information together. To add, update, and delete data easily.
#To organize data in a meaningful way. To retrieve data quickly using keys.

#How do we use Dictionary?

#Syntax: 
#dictionary_name = {"key1": "value1", "key2": "value2", "key3": "value3"}

#When do we use Dictionary? store related information using unique keys.
#Real time examples:
#Situation                   Key          Value
#Student details             Name         Rahul
#Employee details            ID           101
#Mobile details              Brand        Samsung
#Product details             Price        1500
#Bank details                Account-no   12345
```

```python
#Key must be unique. Same values with different keys.
# First way to create dictionary.

#example01
student = {
    "id": 101,
    "name": "moksha",
    "age": 24,
    "age": 25
}
print(student)

#example02
students = {
    "students": 20,
    "students": 21,
    "students": 23,
    "students": 20
}
print(students)

#Second way to create dictionary.

student = dict(name = "moksha", age = 24, course = "python")
print(student)
```

```python
#only add keys
student = dict(name = "moksha", age = 24, course = "python")
print(student.keys())

#only add values
student = dict(name = "moksha", age = 24, course = "python")
print(student.values())

#Methods of Dictionary:-

#update value
student = {
    "id": 101,
    "name": "moksha",
    "age": 24
}
student.update({"age":23})
print(student)

#pop() remove the value
student = {
    "id": 101,
    "name": "moksha",
    "age": 24
}
student.pop("age")
print(student)

#popitem() remove the last inserted item
student = {
    "id": 101,
    "name": "moksha",
    "age": 24
}
student.popitem()
print(student)

#add/update/delete
student = {
    "name": "moksha",
    "age": 24
}
student["city"] = "Hyderabad"
student["name"] = "divya"
del student["age"]

print(student)
```

```python
#Find the maximum value in dictionary

marks = {"divya": 90, "moksha": 70, "shubhangi": 30, "mipsa": 60}
max_value = max(marks.values())
print("Maximum value:", max_value)
```

```python
#Write a Python program to count the total number of key-value pairs in a dictionary.

student = {
    "id": 101,
    "name": "Moksha",
    "age": 24,
    "city": "Hyderabad"
}

count = 0

for key in student:
    count = count + 1

print("Total key-value pairs:", count)
```
