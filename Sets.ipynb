```python
#What is a sets?
#It is an unordered collection of unique elements stored in a single variable.
#It does not allow duplicate values. If we try to add duplicate values, Python automatically keeps only one copy.
#It is mutable, which means we can add or remove elements after creating it.

#Why do we use Sets?
#To store multiple values in a single variable. To remove duplicate values automatically.To store only unique elements.
#To perform mathematical operations such as union, intersection and difference. To check whether an element exists in a collection.

#How do we use Sets? We use curly brackets {} to create a set.

#Syntax: variable_name = {value1, value2, value3}

#When do we use Sets? whenever we want to store unique values and remove duplicate elements.
```

```python
#Example

numbers = {10,20,30,10,30}
print(numbers)

numbers = {10,20,30,10,30}
print(type(numbers))
```

```python
#add()
numbers = {10,20,30,10,30}
numbers.add(40)
print(numbers)

#update()
numbers = {10,20,30,10,30}
numbers.update([70,80,60])
print(numbers)
```

```python
#remove()/discard()
#remove() vs discard() : Remove() raises a keyerror while discard() does nothing no error.

numbers = {10,20,30,10,30}
numbers.remove(10)
numbers.add(400)
print(numbers)

numbers = {10,20,30,10,30}
numbers.discard(20)
numbers.add(25)
print(numbers)

#example01
fruits = {"apple", "banana", "cherry"}

# Using discard() on a missing item does not cause an error
fruits.discard("mango")
print(fruits)  # Output: {'apple', 'banana', 'cherry'}

# Using remove() on a missing item raises a KeyError
fruits.remove("mango")  # KeyError: 'mango'
```

```python
#union() : it combines all the elements

#Combine two variable
A = {1,2,3}
B = {3,4,5}

print(A.union(B))

#Combine three variable
A = {10, 20, 30}
B = {30, 40, 50}
C = {50, 60, 70}

print(A.union(B, C))
```

```python
#intersection() : it can be printed as common elements

A = {10,20,30}
B = {30,40,50}

print(A.intersection(B))
```

```python
#difference() : it can be returns present in A but not in B

A = {4,5,6,7}
B = {7,8,9}

print(A.difference(B))
```

```python
#symmetric_difference() : it returns elements that are present in either of the two sets, but not in both sets.

A = {1,2,3}
B = {3,4,5}

print(A.symmetric_difference(B))
```

```python
#You are given a list of integers. Remove duplicate elements and calculate the sum of unique elements.

A = {60,70,60,50,80,70}
B = set(A)

print(B)
print(sum(B))
```

```python
#You are given three sets. Find the elements common to all three sets.

A = {10, 20, 30}
B = {30, 40, 50}
C = {10, 30, 60}

print(A.intersection(B, C))
```

```python
#You are given two sets of numbers. Find the elements present in the first set but missing in the second set.

A = {10, 20, 30, 40, 50}
B = {30, 40, 60}

print(A.difference(B))
```

```python
# You are given two sets of student roll numbers. One set has subscribed to the English newspaper, 
#and one set has subscribed to the French newspaper. Your task is to find the total number of 
#students who have subscribed to either the English or the French newspaper but not both.

english = set(input("subscribed to english newspaper :"))
french = set(input("subscribed to french newspaper: "))

print("Total students subscribed:- ", english.symmetric_difference(french))
```

```python
#Write a program to calculate the sum of all elements in a list.

numbers = [10, 20, 30, 40, 50]
total = sum(numbers)
print("Sum of all elements:", total)
```
