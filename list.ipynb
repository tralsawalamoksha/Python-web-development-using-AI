```python
#what is list? 
#Collection of multiple items,elements, data stored in single variable. 
#It is mutable, which means we can add, remove, or modify elements after creating it.

#why we use list? 
#To store multiple values in a single variable. Instead of creating separate variables for every value, we can store all values together.
#To add new elements whenever needed.To remove unwanted elements.To modify existing elements.
#To store different types of data in one collection.

# How we use list?  We use square brackets [] to create a list.

#Syntax: variable_name = [value1, value2, value3] #only use square bracket[], not to use this brackets{},()

#when we use list? We use lists whenever we have multiple values that may need to be changed, added, or removed.
```

```python
#Example01
numbers = [10,20,30,40]
print(numbers)

#Example02
names = ["moksha","moksha","mipsa"]
print(names)

#Example03
names = {"moksha","shubhangi","mipsa"}
print(names)

#Example04
details = ("moksha", 20,9.1)
print(details)
```

```python
#Methods of list

#append() add one element at the end of the list
numbers = [10,20,30]
numbers.append(50)
print(numbers)

#extend() add multiple elements.
numbers = [10,20,30]
numbers.extend([40,50,60,70])
print(numbers)

#insert() store index position 1 with name moksha
#syntax : list.insert(insert,value)
names = ["moksha","shubhangi","mipsa"]
names.insert(1,"divya")
print(names)

#remove() paricular value
names = ["moksha","shubhangi","mipsa"]
names.remove("mipsa")
print(names)

#pop() remove using index
numbers = [10,20,30,40]
numbers.pop(2)
print(numbers)

#clear() remove all the elements
numbers = [10,20,30]
numbers.clear()
print(numbers)

#index() findout index position.
names = ["moksha","shubhangi","mipsa","divya"]
print(names.index("divya"))
```

```python
#repeated characters or numbers
name = "programming"

for char in set(name):
    if name.count(char) > 1:
        print(char)
```

```python
#sort() ascending order.low to high
numbers = [50, 20, 90, 10, 80]
numbers.sort()
print(numbers)

#sort(reverse=true) high to low
numbers = [50, 20, 90, 10, 80]
numbers.sort(reverse=True)
print(numbers)
```

```python
#copy(): copy all the elements of list

m = [10, 20, 30, 40, 50]
m.copy()
print(m)
```

```python
#built-in function: It can use built-in directly without creating or defining it ourselves.
#len(): no of elements.
#max(): largest element.
#min(): least element.
#sum(): add numbers.

marks = [50, 20, 90, 10, 80]
print(len(marks))
print(max(marks))
print(sum(marks))
```

```python
#count duplicate elements of list.
marks = [50, 20, 90, 10, 80, 20, 50]
print(marks.count(20))
print(marks.count(50))
```

```python
#Separate even and odd numbers into two different lists by taking 1 to 20 numbers
even = []
odd = []

for i in range(1, 21):
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
#remove an element for index postion 1.

marks = [50, 20, 90, 10, 80]
marks.pop(1)
print(marks)
```

```python
#calculate average of numbers in a list.

numbers = [10, 20, 30, 40, 50]

total = sum(numbers)
count = len(numbers)

average = total / count

print("Sum:", total)
print("Count:", count)
print("Average:", average)
```

```python
#merge two list in one list.

list1 = [10, 20, 30]
list2 = [40, 50, 60]

list1.extend(list2)
print(list1)
```

```python
#find top 3 score marks from 20 members list.

marks = [50, 85, 90, 65, 78, 95, 88, 72, 60, 99, 80, 92, 75, 55, 89, 96, 68, 82, 70, 91]
marks.sort(reverse=True)
print("Top 3 scores:", marks[:3])
```

```python
#findout the how many students have pass and fail based on marks

m = [50, 20, 90, 10, 80, 30, 45, 25, 60, 70]

pass_list = []
fail_list = []

for i in m:
    if i >= 35:
        pass_list.append(i)
    else:
        fail_list.append(i)

print("Pass students marks:", pass_list)
print("Fail students marks:", fail_list)
print("Number of passed students:", len(pass_list))
print("Number of failed students:", len(fail_list))
```
