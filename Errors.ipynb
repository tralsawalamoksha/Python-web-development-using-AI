```python
#In Python, an error is a problem in a program that prevents it from running correctly or produces an unexpected result.
#Python errors are generally classified into three main types:
```

```python
#Syntax Errors : A syntax error occurs when we do not follow Python's rules or grammar while writing code.
#Reason: We forgot to write a colon (:) after the if condition.

#example:
#if 10 > 5
#    print("10 is greater")

#correction:
if 10 > 5:
    print("10 is greater")
```

```python
#Runtime Errors : Runtime errors occur while a program is executing. The code may have correct syntax, but an operation causes an error.

#Common types of runtime errors:

#a) ZeroDivisionError : Occurs when we divide a number by zero.

a = 10
b = 0
print(a / b)

#b) NameError : Occurs when we use a variable that has not been defined.

print(x)

#c) TypeError : Occurs when we perform an operation on incompatible data types.

a = 10
b = "Hello"
print(a + b)

#d) ValueError : Occurs when a function receives the correct type of value but an invalid value.

a = int("hello")

#e) IndexError : Occurs when we access an index that does not exist in a list.

numbers = [10, 20, 30]
print(numbers[5])

#f) KeyError : Occurs when we try to access a dictionary key that does not exist.

student = {"name": "Ram"}
print(student["age"])

#g) FileNotFoundError : Occurs when we try to open a file that does not exist.

file = open("abc.txt", "r")
```

```python
#Logical Errors : 
#Logical errors occur when the program runs without showing an error, but the output is incorrect because of a mistake in the logic.

#example
#a = 10
#b = 20
#result = a * b
#print(result)

#correction
a = 10
b = 20

result = a + b
print(result)
```

```python
#write a program to remove duplicate elements from an array.

n = int(input("Enter your element: "))

arr = []
for i in range(n):
    arr.append(int(input()))

result = []

for i in arr:
    if i not in result:
        result.append(i)

for i in result:
    print(i)
```
