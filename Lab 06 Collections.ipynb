```python
#Find the missing number in a list from 1 to N.

n = 5
list = [1, 2, 4, 5]

total = n * (n + 1) // 2
missing = total - sum(list)

print("Missing number:", missing)
```

```python
#Find the common elements in two lists.

List1 = {10,20,30,40,50,60}
List2 = {30,40,50,70,80,90}

print(List1.intersection(List2))
```

```python
#Remove duplicate elements from a list using set

A = ["moksha", "mipsa", "mipsa", "divya", "shubhangi", "swetha"]
B = list(set(A))

print(B)
```
