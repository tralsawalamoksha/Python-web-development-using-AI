```python
#Write a program to add two dollars and cents.

d1 = int(input("Enter dollars: "))
c1 = int(input("Enter cents: "))

d2 = int(input("Enter dollars: "))
c2 = int(input("Enter cents: "))

total = (d1 * 100 + c1) + (d2 * 100 + c2)

dollars = total // 100
cents = total % 100

print("Total dollars:", dollars)
print("Total cents:", cents)
```

```python
#Write a program to remove duplicate characters from a string.

s = input("Enter a string: ")
r = ""

for char in s:
    if char not in r:
        r = r + char

print("String after removing duplicates:", r)
```

```python
#Write a program to find the minimum occurring character.

s = input("Enter a string: ")

min_char = s[0]

for i in s:
    if s.count(i) < s.count(min_char):
        min_char = i

print("Minimum occurring character:", min_char)
```

```python
#Write a program to find the length of a string without using len().

s = input("Enter a string: ")

count = 0

for i in s:
    count = count + 1

print("Length of the string:", count)
```
