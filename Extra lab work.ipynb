```python
#Write a program to tell the eligibility of the students for the scholarship by providing "Eligible", "Not Eligible", "Partially Eligible"

marks = float(input("Enter student's percentage: "))
income = float(input("Enter annual family income: "))

if marks >= 75 and income <= 200000:
    print("Eligible")
elif marks >= 50 and income <= 400000:
    print("Partially Eligible")
else:
    print("Not Eligible")
```

```python
#Write a program to print the trapezium pattern.

n = int(input())

num = 1
last = n * n + 1

for i in range(n):
    print("--" * i, end="")

    for j in range(n - i):
        print(num, end="")
        num = num + 1
        if j < n - i - 1:
            print("*", end="")

    for j in range(n - i):
        print("*" if j == 0 else "", end="")
        print(last, end="")
        last = last + 1

    last = last - 2 * (n - i)
    print()
```
