```python
#What is strings? 
#It is a collection of characters enclosed within single quotes (' '), double quotes (" ") or triple quotes (''' ''' or """ """).
#Used to store text data such as names, addresses, messages, and sentences.
#It is immutable, which means we cannot modify its individual characters after creation.

#Why we use strings? 
#To store text data like names, addresses, & phone numbers.
#To store messages & sentences. To display information to users.
#To perform text operations such as searching, replacing, and joining. To process user input.

#How we use strings? We use single quotes, double quotes, or triple quotes to create strings.

#Syntax: variable_name = "value1"

#When do we use strings?
#We use strings whenever we want to store, process, or display text information.

#Real-world examples: Storing student names, Storing employee names, Storing email addresses, Storing passwords, Storing customer addresses.
#Displaying messages and notifications.
#Reading user input.
```

```python
# findout y is in which index position
name = " python"
print (name.find("y"))
print(name[3]) #postive index
print(name[-1]) #negative index
```

```python
#slicing - dividing the parts
#string_name[start:end]

name = "moksha"
print(name[0:4]) #output 01

#slicing from the beginning

name = "python"
print(name[:5])

#slicing from the end
name = "developnment"
print(name[3:])

#negative index positon
name = "security"
print(name[-3:-1])

#print reverse of the case
name = "moksha"
print(name[::-1]) #backward direction, move backward
print(name[::1])  #forward direction, move forward
print(name[::2])  #skip 1 character
print(name[::3]) #skip 2 character
```
