```python
#write a program to find all special numbers between 2 limits m and n (both inclusive)? Assume that m and n are two-digit numbers.

m = int(input("Enter a number: "))
n = int(input("Enter a number: "))

while m <= n:
    a = m//10
    b = m%10
  
    s = (a + b) + (a * b) 
  
    if m==s:
        print(m)

    m = m  + 1
```

```python
#Write a program to check whether a given number is a strong number or not.
n = int(input())
temp = n
sum = 0

while temp > 0:
    digit = temp % 10
    fact = 1
    i = 1

    while i <= digit:
        fact = fact * i
        i = i + 1

    sum = sum + fact
    temp = temp // 10

if sum == n:
    print("Yes, it's a strong number")
else:
    print("No, it's not a strong number")
```

```python
#Write an algorithm to find all the prime numbers from 2 to the given positive number.

num = int(input("Enter a number: "))
count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count = count + 1

if count == 2:
    print("It is a prime number")
else:
    print("It is not a prime number")
```
