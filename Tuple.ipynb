```python
#What is tuple?  
#It can be defined as the ordered collection of elements. It allows dulplicate values but can not modified after creation. It is immutable.

#Why we use tuples? 
#To store multiple values, To allow duplicate values, To protect data, To store fixed data, Better performance

#How we use tuples?
#We use parentheses () to create a tuple, and elements are separated by commas.

#Syntax: variable_name = (value1, value2, value3)

#When do we use Tuples?
#We use tuples when we want to store multiple values that should not be changed after creation.

#Mutable: Can be modified vs Immutable: Cannot be modified
```

```python
#Example
student = ("moksha", 107, "python", 1233445566)
print(student)
print(type(student))

#modified the value
numbers = [10,20,30]
print(numbers)

numbers[0] = 100
print(numbers)
```

```python
#count numbers how many time it is repeated
student = (10,20,30,40)
result = student.count(10)
print(result)

#specify the index position
numbers = (10,20,30,40,50,60,70)
result = numbers.index(50)
print(result)
```

```python
m = (("moksha", 90), ("shubhangi", 80), ("mipsa", 60))

total = 0

for student in m:
    total = total + student[1]

print(m)
print("Total marks:", total)
```
